# Adjacency matrix

Types 4:
- Unweighted (show connections with 1)
    - Undirected (u to v, v to u)
    - Directed (u to v)

- Weighted (show connections with edge weight)
    - Undirected (u to v, v to u)
    - Directed (u to v)

### Adjacency matrix for Undirected Unweighted Graph

matrix_undirected_unweighted_p1.png

matrix_undirected_unweighted_p2.png

##### Python

```python
import numpy as np

class Graph:

    def __init__(self, num_vertices):
        """Initialize the graph with the given number of vertices."""

        self.num_vertices = num_vertices   # Total number of vertices

        # Create a V x V matrix filled with 0
        # 0 → no edge, 1 → edge exists
        self.adj_matrix = np.zeros((num_vertices, num_vertices), dtype=int)


    def add_edges(self, edges):
        """Add edges to the graph (UNDIRECTED)."""

        # edges format: [(source, destination), ...]
        for source, destination in edges:

            # For UNDIRECTED graph:
            # Add edge in BOTH directions
            self.adj_matrix[source][destination] = 1
            self.adj_matrix[destination][source] = 1


    def print_matrix(self):
        """Print the adjacency matrix."""

        print("Adjacency Matrix:")

        # Traverse each row
        for row in self.adj_matrix:
            print(row)   # Each row represents connections from that vertex


    def find_neighbors(self, vertex):
        """Find and print neighbors of a given vertex."""

        neighbors = []   # Temporary list to store neighbors

        # Traverse row = vertex
        for i in range(self.num_vertices):

            # If there is an edge vertex — i
            if self.adj_matrix[vertex][i] == 1:
                neighbors.append(i)

        # Print neighbors (space-separated)
        print(f"\nNeighbors of vertex {vertex}: ", end="")

        for i in range(len(neighbors)):
            print(neighbors[i], end="")

        print()  # New line
```

### Adjacency matrix for Directed Weighted graph

Matrix_W_directed_p1.png

Matrix_W_directed_p2.png

##### Python

```python
import numpy as np

class Graph:

    def __init__(self, num_vertices):
        """Initialize the graph with the given number of vertices."""

        self.num_vertices = num_vertices   # Total number of vertices

        # Create a V x V matrix filled with 0
        # 0 → no edge, non-zero → weight of edge
        self.adj_matrix = np.zeros((num_vertices, num_vertices), dtype=int)


    def add_edges(self, edges):
        """Add edges with weights to the graph (DIRECTED)."""

        # edges format: [(source, destination, weight), ...]
        for source, destination, weight in edges:

            # For DIRECTED graph:
            # Store weight only for source → destination
            self.adj_matrix[source][destination] = weight


    def print_matrix(self):
        """Print the adjacency matrix."""

        print("Adjacency Matrix:")

        # Traverse each row
        for row in self.adj_matrix:
            print(row)   # Each row shows outgoing edges with weights


    def find_neighbors(self, vertex):
        """Find and print neighbors of a given vertex."""

        neighbors = []   # Temporary list to store neighbors

        # Traverse row = vertex
        for i in range(self.num_vertices):

            # If weight is non-zero → edge exists
            if self.adj_matrix[vertex][i] != 0:
                neighbors.append(i)

        # Print neighbors with comma separation
        print(f"\nNeighbors of vertex {vertex}: ", end="")

        for i in range(len(neighbors)):
            print(neighbors[i], end="  ")   # Extra space for readability

        print()  # New line
```

### Adjacency list for Undirected unWeighted graph

#### Python

```python
class Edge:
    def __init__(self, destination, next_node=None):
        """Edge to represent a connection from source to destination."""
        self.destination = destination
        self.next = next_node
###############################################
class AdjacencyListGraph:

    def __init__(self, num_vertices):
        """Initialize the graph with the given number of vertices."""

        self.num_vertices = num_vertices   # Total number of vertices

        # Create a list of size = num_vertices
        # Each index will store the head of a linked list
        self.adj_list = [None] * num_vertices


    def add_edges(self, edges):
        """Add multiple edges to the graph (UNDIRECTED)."""

        # edges format: [(source, destination), ...]
        for source, destination in edges:

            # For UNDIRECTED graph:
            # Add edge in BOTH directions
            self.add_edge(source, destination)      # source → destination
            self.add_edge(destination, source)      # destination → source


    def add_edge(self, source, destination):
        """Helper method to add a single edge."""

        # Step 1: Create a new edge node
        new_edge = Edge(destination)

        # Step 2: Insert at the beginning (prepend)
        # This ensures O(1) insertion time
        new_edge.next = self.adj_list[source]

        # Step 3: Update the head of the list
        self.adj_list[source] = new_edge


    def print_list(self):
        """Print the full adjacency list."""

        print("Adjacency List:")

        # Traverse each vertex
        for i in range(self.num_vertices):

            print(f"{i}: ", end="")

            # Traverse linked list of edges for vertex i
            current = self.adj_list[i]   # head node

            while current:

                # Print neighbor (destination only, since unweighted)
                print(f"{current.destination}", end=" ")

                current = current.next  # Move to next edge

            print()  # Move to next line


    def find_neighbors(self, vertex):
        """Print all neighbors of a given vertex."""

        print(f"\nNeighbors of vertex {vertex}: ", end="")

        current = self.adj_list[vertex]

        # Traverse all neighbors
        while current:

            # Print current neighbor
            print(f"{current.destination}", end="")

            # Add comma if not the last element
            if current.next:
                print(",", end="")

            current = current.next  # Move to next edge

        print()  # New line
```

### Adjacency list for Directed unWeighted graph

##### Python

```python
class Edge:
    def __init__(self, destination, next_node=None):
        self.destination = destination  # Ending vertex (neighbor)
        self.next = next_node       # Pointer to next edge (linked list)


class AdjacencyListGraph:

    def __init__(self, num_vertices):
        self.num_vertices = num_vertices   # Total number of vertices

        # Create a list of size = num_vertices
        # Each index will store the head of a linked list
        self.adj_list = [None] * num_vertices


    def add_edges(self, edges):
        """Add multiple edges to the graph (DIRECTED)."""

        # edges format: [(source, destination), ...]
        for source, destination in edges:

            # For DIRECTED graph:
            # Add edge in only ONE direction
            self.add_edge(source, destination)


    def add_edge(self, source, destination):
        """Helper method to add a single edge."""

        # Step 1: Create a new edge node
        new_edge = Edge(destination)

        # Step 2: Insert at the beginning (prepend)
        # This ensures O(1) insertion time
        new_edge.next = self.adj_list[source]

        # Step 3: Update the head of the list
        self.adj_list[source] = new_edge


    def print_list(self):
        """Print the full adjacency list."""

        print("Adjacency List:")

        # Traverse each vertex
        for i in range(self.num_vertices):

            print(f"{i}: ", end="")

            # Traverse linked list of edges for vertex i
            current = self.adj_list[i]

            while current:

                # Print only destination (unweighted graph)
                print(current.destination, end=" ")

                current = current.next  # Move to next edge

            print()  # Move to next line


    def find_neighbors(self, vertex):
        """Print all neighbors of a given vertex."""

        print(f"\nNeighbors of vertex {vertex}: ", end="")

        current = self.adj_list[vertex]

        # Traverse all neighbors
        while current:

            # Print current neighbor
            print(f"{current.destination}", end="")

            # Add comma if not the last element
            if current.next:
                print(",", end="")

            current = current.next  # Move to next edge

        print()  # New line
```

### Adjacency list for Directed weighted graph

Wedge.png

W_directed_weighted.png

W_directed_weighted2.png

##### Python

```python
class WEdge:
    def __init__(self, destination, weight, next_node=None):
        self.destination = destination
        self.weight = weight
        self.next = next_node
###############################################################
class AdjacencyListGraph:

    def __init__(self, num_vertices):
        self.num_vertices = num_vertices   # Total number of vertices
        self.adj_list = [None] * num_vertices
        # Create a list of size = num_vertices
        # Each index will store the head of a linked list of edges


    def add_edges(self, edges):
        """Add multiple edges to the graph (DIRECTED)."""

        # edges format: [(source, destination, weight), ...]
        for source, destination, weight in edges:

            # Add edge from source → destination
            # (Only one direction since it is DIRECTED)
            self.add_edge(source, destination, weight)


    def add_edge(self, source, destination, weight):
        """Helper method to add a single edge with weight."""

        # Step 1: Create a new edge node
        new_edge = WEdge(destination, weight)

        # Step 2: Insert at the beginning (prepend)
        # This ensures O(1) insertion time
        new_edge.next = self.adj_list[source]

        # Step 3: Update the head of the list
        self.adj_list[source] = new_edge


    def print_list(self):
        """Print the full adjacency list."""

        print("Adjacency List with Weights:")

        # Traverse each vertex
        for i in range(self.num_vertices):

            print(f"{i}: ", end="")

            # Traverse linked list of edges for vertex i
            current = self.adj_list[i]
            print("s=", i, end="  ")

            while current:

                # Print (destination, weight)
                print(f"(d={current.destination}, w={current.weight})", end="")

                # Add comma if not last edge
                if current.next:
                    print(", ", end="")

                current = current.next  # Move to next edge

            print()  # Move to next line after each vertex


    def find_neighbors(self, vertex):
        """Print all neighbors of a given vertex."""

        print(f"\nNeighbors of vertex {vertex}: ", end="")

        current = self.adj_list[vertex]

        # Traverse all outgoing edges from this vertex
        while current:

            # Print (destination, weight)
            print(f"({current.destination}, {current.weight})", end="")

            # Add comma if not last
            if current.next:
                print(", ", end="")

            current = current.next  # Move to next edge

        print()  # New line
```
