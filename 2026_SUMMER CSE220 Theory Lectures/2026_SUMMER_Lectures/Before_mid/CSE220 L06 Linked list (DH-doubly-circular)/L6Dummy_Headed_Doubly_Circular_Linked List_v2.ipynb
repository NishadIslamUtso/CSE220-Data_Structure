## Head and Tail in a Doubly Linear Linked List

- **Head:** The first node of the list.
  - `head.prev = null`

- **Tail:** The last node of the list.
  - `tail.next = null`
  - Accessing the tail takes **O(n)** time if no tail reference is given.

---

## Head and Tail in a Doubly Circular Linked List

- **Head:** The first node of the list.
  - `head = tail.next` (can be accessed in this way)

- **Tail:** The last node of the list.
  - Accessing the tail takes **O(1)** time.
  - `tail = head.prev` (can be accessed in this way)

## Linked List Code Shortcuts (Remember!! Important!!)

- **Loop termination condition:**
  - **Linear Linked List:** Stop when the node becomes `null`.
  - **Circular Linked List:** Stop when the node reaches `head`.

- **Starting point of traversal:**
  - **Dummy-headed Linked List:** Start from `head.next`, because the dummy head does not store data.
  - **Non-dummy-headed Linked List:** Start from `head`, because the head stores data.

## 2.5.1 LL Creation:
It is similar to creating a non-dummy singly linaer linked list. The difference is that we need to ensure the connections in a proper way.
The below function takes an array and creates a dummy-headed doubly circular linked list.

````python
def printLL_forward(head):
    print("Forward Printing: ", end='')

    current_node = head.next # since dummy headed, value starts after head
    while current_node!=head: # end at head
        print(current_node.elem, end=' -> ')
        current_node = current_node.next # i= i+1
    print("END")
````

````python
import numpy as np

def create_doubly_circular_LL(arr):
    # State 1 of the image above
    head = Node(None)  # Dummy head node

    tail = head # here, current_node is tail
    for i in range(0, len(arr), 1):
        #----Code of append-----
        newNode = Node(arr[i])

        tail.next = newNode  # Link current node to new node
        newNode.prev = tail  # Link new node back to current node---extra

        tail = newNode  # tail=tail.next Move to the new node
        #-----------------------

    # Making it circular by connecting the last node back to the head--extra
    tail.next = head
    head.prev = tail

    return head  # Returning the head of the newly created linked list


arr = np.array([10, 20, 30, 40])
head2 = create_doubly_circular_LL(arr)

# printing the newly created linked list
printLL_forward(head2)
print('-'*50)

arr = np.array([])
head22 = create_doubly_circular_LL(arr)

# printing the newly created linked list
printLL_forward(head22)
print('-'*50)

arr = np.array([10])
head22 = create_doubly_circular_LL(arr)

# printing the newly created linked list
printLL_forward(head22)
````

## 2.5.3 Size:
- In order to count the number of nodes present in a linked list, we just need to count how many times the iteration process is running during a traversal.
- **ignore the dummy head, while counting the nodes**
- Forward or Backward traversal--does not matter

size.png

get_size.png

````python
def get_size(head):
    if head.next == head.prev:  # If the list is empty, return 0
        return 0

    count = 0
    current_node = head.next  # Start from the first element after head
    while current_node != head:  # Loop until we circle back to head
        count += 1
        current_node = current_node.next
    return count


arr = np.array([])
head5 = create_doubly_circular_LL(arr)

# printing the newly created linked list
printLL_forward(head5)
print("Size:", get_size(head5))
print('-'*50)

arr = np.array([10, 20, 30, 40, 50])
head3 = create_doubly_circular_LL(arr)

# printing the newly created linked list
printLL_forward(head3)
print("Size:", get_size(head3))
print('-'*50)

arr = np.array([10, 20, -30])
head4 = create_doubly_circular_LL(arr)

# printing the newly created linked list
printLL_forward(head4)
print("Size:", get_size(head4))
print('-'*50)
````

## 2.5.5 Insertion:
- To insert a new node in the list, you need the reference to the predecessor to link in the new node.
- Unlike for a singly-linked linear list, **there is no "special" case here**, since there is always a valid predecessor node available, thanks to the dummy head.
- **Head does not need to change.**
#### -
- Time complexity O(n)
- Space complexity O(1)

````python
def insertAt(head, given_val, idx):
    size = get_size(head)
    if idx < 0 or idx > size:
        print("Wrong Index. Insertion Failed!")
        return None

    newNode = Node(given_val)

    # Special case for inserting at the beginning , Now the head will be precessor
    if idx == 0:
        predecessor = head
    else:
        # General case for inserting at idx (other than the start)
        predecessor = getNode(head, idx - 1)
    successor = predecessor.next

    # Establishing the connection of newNode
    newNode.next = successor
    newNode.prev = predecessor

    # Connecting rest of links
    predecessor.next = newNode
    successor.prev = newNode##--condition


arr = [10, 20, 30, 40]
head3 = create_doubly_circular_LL(arr)
print("Original List:")
printLL_forward(head3)

# Insert 5 at index 0--HERE HEAD IS PRED
insertAt(head3, 5, 0)
print("\nList after inserting 5 at index 0 (just after head):")
printLL_forward(head3)


# Insert 15 at index 1
insertAt(head3, 15, 1)
print("\nList after inserting 15 at index 1 (middle):")
printLL_forward(head3)

# Insert 35 at index 3
insertAt(head3, 35, 3)
print("\nList after inserting 35 at index 3 (middle):")
printLL_forward(head3)

# Insert 50 at index 3
insertAt(head3, 50, 7)
print("\nList after inserting 50 at index 7 (after tail):")
printLL_forward(head3)
````

# 2.5.6 Removal:
Removing an element from the list is done by removing the node that contains the element.
Just like inserting a new node in a list, removing requires that you have the reference to the predecessor node.

Since we're using a doubly-linked list, finding a predecessor of a node is trivial — it's n.prev. And thanks to the dummy head, there is no "special" case here as well.

- Time complexity O(n)
- Space complexity O(1)

````python
def removeAt(head, idx):
    size = get_size(head)
    if idx < 0 or idx >= size:
        print("Wrong Index. Deletion Failed!")
        return

    remNode = getNode(head, idx)
    predecessor = remNode.prev #predecessor= getNode(head, idx - 1) #---BOTH FINE
    successor = predecessor.next.next  # successor = remNode.next #---BOTH FINE

    # Connecting rest of links
    predecessor.next = successor
    successor.prev = predecessor ####--condition

    # Removing the connections of remNode
    remNode.next = remNode.prev = None


arr = [10, 20, 30, 40, 50]
head3 = create_doubly_circular_LL(arr)
print("Original List:")
printLL_forward(head3)

# Remove node at index 0
removeAt(head3, 0)
print("\nList after removing node at index 0 (Just after head):")
printLL_forward(head3)
print('-'*50)

# Remove node at index 2
removeAt(head3, 2)
print("\nList after removing node at index 2 (middle):")
printLL_forward(head3)
print('-'*50)

# Remove node at last index (now index 2 after removals)
removeAt(head3, 2)
print("\nList after removing node at last index 2 (tail):")
printLL_forward(head3)
print('-'*50)


# Remove node at wrong index
removeAt(head3, 5)
print("\nList after removing node at 5 (wrong index)")
printLL_forward(head3)
print('-'*50)

# Remove node at wrong index
removeAt(head3, -3)
print("\nList after removing node at -3 (wrong index)")
printLL_forward(head3)
print('-'*50)
````

## Out-of-place-reversal [Easy]
- craetes a new copy of original LL

reverse_out_of_place.png

````python
def reverseLL_out_of_place(head):
    """Reverses the doubly dummy-headed circular linked list using an out-of-place approach."""
    if head.next == head:  # Empty list
        print("List is empty. Nothing to reverse.")
        return


    temp_tail = head.prev  # Start from the last element before head
    rev_head = Node(None)  # New dummy head
    rev_tail = rev_head # Pointer for building reversed list

    while temp_tail != head:
        #---append code-----
        newNode = Node(temp_tail.elem)

        rev_tail.next = newNode
        newNode.prev = rev_tail

        rev_tail = newNode
        #---------------------------
        temp_tail = temp_tail.prev

    # Making the reversed list circular
    rev_tail.next = rev_head
    rev_head.prev = rev_tail

    print("List reversed successfully.")

    return rev_head


# Test the reverseLL_out_of_place function
arr = [10, 20, 30, 40, 50]
head = create_doubly_circular_LL(arr)
print("Original List:")
printLL_forward(head)
printLL_backward(head)

# Reverse the list
reverseLL_out_of_place(head)

print("\nList after reversal:")
printLL_forward(head)
printLL_backward(head)
````

````python
def reverseLL(head):
    """Reverses the doubly dummy-headed circular linked list."""
    if head.next == head:  # Empty list
        print("List is empty. Nothing to reverse.")
        return

    current_node = head
    while True:
        # Use a temporary variable to hold the next reference
        temp = current_node.next
        # Swap next and prev manually
        current_node.next = current_node.prev
        current_node.prev = temp

        # Move to the next node (which was previously prev)
        current_node = current_node.prev
        if current_node == head:  # Completed one full cycle
            break

    print("List reversed successfully.")


# Test the reverseLL function
arr = [10, 20, 30, 40, 50]
head = create_doubly_circular_LL(arr)
print("Original List:")
printLL_forward(head)
printLL_backward(head)

# Reverse the list
reverseLL(head)

print("\nList after reversal:")
printLL_forward(head)
printLL_backward(head)
````

#  Dummy-headed doubly circular Linked List Class - optional
Let's try to combine all the codes above and try to create a DHDCLL(Dummy headed doubly circular Linked List) Class

````python
# Defining the Node class
class Node:
    def __init__(self, elem, next=None, prev=None):
        self.elem = elem  # The element that we want to put in the list
        self.next = next  # The reference to the next node in the list
        self.prev = prev  # The reference to the previous node in the list

#____________________________________________________________________
# Defining the DHDCLL class
class DHDCLL:
    def __init__(self):
        self.head = Node(None)  # Dummy head node
        self.head.next = self.head
        self.head.prev = self.head

    def create_doubly_circular_LL(self, arr):
        current_node = self.head
        for i in range(0, len(arr), 1):
            newNode = Node(arr[i])
            current_node.next = newNode
            newNode.prev = current_node
            current_node = newNode
        # Making it circular
        current_node.next = self.head
        self.head.prev = current_node

    def printLL_forward(self):
        print("Forward Printing: ", end='')
        current_node = self.head.next
        while current_node != self.head:
            print(current_node.elem, end=' -> ')
            current_node = current_node.next
        print()

    def printLL_backward(self):
        print("Backward Printing: ", end='')
        current_node = self.head.prev
        while current_node != self.head:
            print(current_node.elem, end=' <- ')
            current_node = current_node.prev
        print()

    def get_size(self):
        if self.head.next == self.head:  # Empty list case
            return 0
        count = 0
        current_node = self.head.next
        while current_node != self.head:
            count += 1
            current_node = current_node.next
        return count

    def getNode(self, target_index):
        current_node = self.head.next
        current_index = 0
        while current_node != self.head:
            if current_index == target_index:
                return current_node
            current_index = current_index + 1
            current_node = current_node.next
        return None  # Return None if index is out of bounds

    def insertAt(self, given_val, idx):
        size = self.get_size()
        if idx < 0 or idx > size:
            print("Wrong Index. Insertion Failed!")
            return None

        newNode = Node(given_val)
        # Special case for inserting at the beginning
        if idx == 0:
            predecessor = self.head
        else:
            # General case for inserting at idx
            predecessor = self.getNode(idx - 1)

        successor = predecessor.next
        # Establishing connections for the newNode
        newNode.next = successor
        newNode.prev = predecessor
        predecessor.next = newNode
        successor.prev = newNode
        return newNode

    def removeAt(self, idx):
        size = self.get_size()
        if idx < 0 or idx >= size:
            print("Wrong Index. Deletion Failed!")
            return

        remNode = self.getNode(idx)
        predecessor = remNode.prev
        successor = remNode.next

        # Connecting remaining links
        predecessor.next = successor
        successor.prev = predecessor

        # Isolating the removed node
        remNode.next = remNode.prev = None
        return remNode  # Return removed node if needed

    def reverseLL(self):
        """Reverses the doubly dummy-headed circular linked list."""
        if self.head.next == self.head:  # Empty list
            print("List is empty. Nothing to reverse.")
            return

        current_node = self.head
        while True:
            # Use a temporary variable to hold the next reference
            temp = current_node.next
            # Swap next and prev manually
            current_node.next = current_node.prev
            current_node.prev = temp

            # Move to the next node (which was previously prev)
            current_node = current_node.prev
            if current_node == self.head:  # Completed one full cycle
                break
        print("List reversed successfully.")

    def reverseLL_out_of_place(self):
        if self.head.next == self.head:
            print("List is empty. Nothing to reverse.")
            return

        temp_tail = self.head.prev
        rev_head = Node(None)
        rev_tail = rev_head

        while temp_tail != self.head:
            newNode = Node(temp_tail.elem)
            rev_tail.next = newNode
            newNode.prev = rev_tail
            rev_tail = newNode
            temp_tail = temp_tail.prev

        rev_tail.next = rev_head
        rev_head.prev = rev_tail
        self.head = rev_head
        print("List reversed successfully.")
````
